```python
sub_ensemble=pd.read_csv('../data/geochron/sub_ensemble.csv',header=0)# ensemble of MCMC 
sub_mu=pd.read_csv('../data/geochron/sub_mu.csv')# Subsidence amount mean
mean_age=pd.read_csv('../data/geochron/sub_mean.csv')# Mean age interpolated 
sub_params=pd.read_csv('../data/geochron/sub_params.csv') # the beta and t naught
```

```python
plt.figure(figsize=(10,6))

for i,val in sub_ensemble.iterrows():
    plt.plot(mean_age,val,color='b',alpha=0.005)

plt.plot(mean_age,sub_mu,color='black')
    
plt.ylim(0,4000)
plt.gca().invert_yaxis()
plt.gca().invert_xaxis()


```

```python
beta_median=sub_params['beta'].median()
t0=sub_params['t_s'].median()
```

```python
sns.histplot(sub_params['beta'],bins=50,)
plt.vlines(beta_median,ymin=0,ymax=19000,color='black',linewidth=3,linestyles='dashed')

plt.xlabel(r'$\beta$')
plt.ylabel('Frequency')

plt.xlim(1,4)
plt.ylim(0,19000)
```

```python
sns.histplot(sub_params['t_s'],bins=50,color='red')
plt.vlines(t0,ymin=0,ymax=18000,color='black',linewidth=3,linestyles='dashed')

plt.xlabel(r'$T_0$ (Ma)')
plt.ylabel('Frequency')


plt.ylim(0,17000)

plt.gca().invert_xaxis()
```
