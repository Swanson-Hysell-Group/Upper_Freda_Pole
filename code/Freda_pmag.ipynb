## BR section E/I

In this section we will be using the f-factor determinedd from the SVEI method applied to the Wolf Pt. dataset for the following reasons:

- WP has the highest n for any single section
- There is no vertical axis rotation
- It has the same lithologies as specimens from other sections 

We will apply a single flattening factor for the fisher distributed poles and then propagate the flattening uncertainty to the kent mean pole 
using the svei_

```python
# This function is for reporting the flattened fisher mean, unflattened with bounds
def unsquish_fisher_print(data,med,low,hi):

    flat_fish=ipmag.fisher_mean(dec=data['dir_dec'].tolist(),inc=data['dir_inc'].tolist())
    print('The flattened fisher mean is :')
    ipmag.print_direction_mean(flat_fish)
    
    # now get the fisher mean of the median 
    unflat_med=ipmag.unsquish(data['dir_inc'],med)
    unflat_med_mean=ipmag.fisher_mean(dec=data['dir_dec'].tolist(),unflat_med)
    print(f'Fisher mean for an f-factor of {med}:')
    ipmag.print_direction_mean(unflat_med_mean)
```

In this section: 
- we first get the lower and upper confidence interval from the output of the resampled paleolatitudes 
- we then calculate the paleolatitude as implied by the unflattened fishermean pole 
- these data are then used for the latitudinal uncertainty

```python
def lat_range(plats):
 plat_upper, plat_lower = abs(np.round(np.percentile(plats, [2.5, 97.5]), 1))
 plat_median=abs(np.median(plats))
 lat_uncert=np.array([[abs(plat_median-plat_lower)],[abs(plat_upper-plat_median)]])

 return(lat_uncert)
```

```python
wp_lat_range=lat_range(wp_plats)
lr_lat_range=lat_range(lr_plats)
wrf_lat_range=lat_range(wrf_plats)
br_lat_range=lat_range(br_plats)
henry_lat_range=lat_range(henry_plats)
```

```python

n = 20000

# UF_resample = ipmag.kentrot(kent_distribution_95, n=n)
# LF_resample = ipmag.kentrot(henry_kent , n=n)

UF_resample=ipmag.fisher_mean_resample(wp_unflat_HT_pole['alpha95'],n=n,dec=wp_unflat_HT_pole['dec'],
                                       inc=wp_unflat_HT_pole['inc'])

LF_resample=ipmag.fisher_mean_resample(Henry_ht_pole['alpha95'],n=n,dec=Henry_ht_pole['dec'],
                                       inc=Henry_ht_pole['inc'])


wp_lat, wp_lon = 47.0, -89.030556

LF_plats =np.empty(n)
UF_plats = np.empty(n)
L_F_distances = np.empty(n)
L_F_rates = np.empty(n)
L_F_gc_rate=np.empty(n)

for i in range(n):

    LF_plats[i] = 90 - pmag.angle(LF_resample[i], [wp_lon, wp_lat])[0]
    UF_plats[i] = 90 - pmag.angle(UF_resample[i], [wp_lon, wp_lat])[0]

    L_F_distances[i] = pmag.angle(LF_resample[i], UF_resample[i])[0]
    L_F_gc_rate[i]=L_F_distances[i]/abs(lf_post[i]-uf_post[i])
    L_F_rates[i] = (LF_plats[i] - UF_plats[i]) / abs(lf_post[i] - uf_post[i]) * 110.567 / 10


```
